```python
# Dadas estas listas, crea un solo objeto de la forma:

#       'annotations': [
#           {'video_id': '001', video_name: 'columpioscam...', cam_id: 'columpios1', 'label': 'playing_together'}, 
#           {'video_id': '002', video_name: 'columpioscam...', cam_id: 'columpios3', 'label': 'waiting'}, 
#       ],

# De las listas, modifica video_id para llamarse video_name, y video_id hazlo solo números como en el ejemplo


# --- Leer el contenido del archivo ---
with open("annotations_raw.txt", "r", encoding="utf-8") as f:
    raw = f.read()

# --- Ejecutar el contenido para crear las variables columpios1, columpios3, columpios4 ---
exec(raw)

# Ahora las variables columpios1, columpios3, columpios4 existen en el entorno

# --- Unir todas las listas ---
all_lists = {
    'columpios1': columpios1,
    'columpios2': columpios1,
    'columpios3': columpios3,
    'columpios4': columpios4,
    'hundido4': hundido4,
    'hundido5': hundido5,
    'hundido7': hundido7
}
```

```python

# --- Código para fusionar ---

""" annotations = []
counter = 1

for cam_id, videos in all_lists.items():
    for v in videos:
        annotations.append({
            'video_id': f"{counter:03d}",  # formato 001, 002, 003...
            'video_name': v['video_id'],   # cambia nombre del campo
            'cam_id': cam_id,
            'label': v['label']
        })
        counter += 1
 """

 # --- Código para fusionar con mapeo de clases a 2 categorías ---

annotations = []
counter = 1

for cam_id, videos in all_lists.items():
    for v in videos:
        
        # Mapeo de la clase original a Transit / Social
        original_label = v['label']
        mapped_label = "Transit" if original_label == "Transit" else "Social"

        annotations.append({
            'video_id': f"{counter:03d}",     # 001, 002, 003...
            'video_name': v['video_id'],      # cambia nombre del campo
            'cam_id': cam_id,
            'label': mapped_label             # <--- mapeo aquí
        })
        
        counter += 1

```

#### Splits

```python
import random

# --- Crear splits (determinista) ---
random.seed(42)  # ← Siempre el mismo resultado
ids = [a['video_id'] for a in annotations]
random.shuffle(ids)

split_index = int(len(ids) * 0.8)
train_ids = ids[:split_index]
val_ids = ids[split_index:]
```

```python
final_object = {
    'annotations': annotations,
    'split': {
        'train': train_ids,
        'eval': val_ids
    }
}
```

```python
import pickle

# --- Guardar como .pkl ---
with open("annotations.pkl", "wb") as f:
    pickle.dump(final_object, f)

print("Archivo 'annotations.pkl' creado con éxito.")
```
